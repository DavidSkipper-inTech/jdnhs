```python
#staff group dim table
silver_workforce = (spark.table("silver_workforce_fte"))
silver_workforce.show(10)
```

```python
from pyspark.sql.functions import col, lower,trim, row_number
from pyspark.sql.window import Window

window_spec = Window.orderBy("staff_group")

workforce_staff_group = (silver_workforce
      
    .select(col("staff_group"))
    .distinct()  
    .withColumn("staff_group_key", row_number().over(window_spec)) 
) 
workforce_staff_group = workforce_staff_group.select( "staff_group_key", "staff_group")

workforce_staff_group.show()
```

```python
(workforce_staff_group.write \
    .format("delta") \
    .mode("overwrite") \
    .option(
        "path",
        "abfss://gold@jdnhsbronze.dfs.core.windows.net/dim_staff/"
    ) \
    .saveAsTable(
        "staff_dimension"
    ))
```
