```python
# dim region gold

silver_workforce = (spark.table("silver_sick_leave"))
silver_workforce.show(10)





```

```python
from pyspark.sql.functions import col, row_number
from pyspark.sql.window import Window

window_spec = Window.orderBy("region_name")

dim_region_gold = (silver_workforce
.select(col("region_name"))
.distinct()
.withColumn( "region_key", row_number().over(window_spec))
.select("region_key", "region_name"))
dim_region_gold.show()
```

```python


# write table to gold schema
(dim_region_gold.write \
    .format("delta") \
    .mode("overwrite") \
    .option(
        "path",
        "abfss://gold@jdnhsbronze.dfs.core.windows.net/dim_region/"
    ) \
    .saveAsTable(
        "region_dimension"
    ))
```
